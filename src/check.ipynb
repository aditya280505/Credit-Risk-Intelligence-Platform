```python
import pandas as pd
from pathlib import Path
from sklearn.model_selection import train_test_split

# ==================================================
# LOAD FEATURE-ENGINEERED DATA
# ==================================================

BASE_DIR = Path.cwd().parent

DATA_PATH = (
    BASE_DIR
    / "data"
    / "processed"
    / "loan_feature_engineered.csv"
)

df = pd.read_csv(DATA_PATH)

print("✅ Dataset loaded")
print("Shape:", df.shape)


# ==================================================
# SEPARATE FEATURES AND TARGET
# ==================================================

X = df.drop(columns=["default"])
y = df["default"]


# ==================================================
# SAME TRAIN / TEST SPLIT
# ==================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


print("\n✅ Test data created")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python

# ==================================================
# APPLY SCALER + CALCULATE ACTUAL MODEL METRICS
# ==================================================

from sklearn.metrics import accuracy_score, f1_score, roc_auc_score
import pandas as pd

# Scale test data
X_test_scaled = scaler.transform(X_test)

print("✅ Test data scaled")
print("Scaled test shape:", X_test_scaled.shape)


# ==================================================
# MODELS
# ==================================================

models = {
    "Logistic Regression": logistic_model,
    "Random Forest": rf_model,
    "XGBoost": xgb_model
}


# ==================================================
# CALCULATE METRICS
# ==================================================

results = []

for name, model in models.items():

    print(f"\n⏳ Evaluating {name}...")

    y_pred = model.predict(X_test_scaled)

    if hasattr(model, "predict_proba"):

        y_prob = model.predict_proba(
            X_test_scaled
        )[:, 1]

    else:

        y_prob = model.decision_function(
            X_test_scaled
        )

    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    f1 = f1_score(
        y_test,
        y_pred
    )

    roc_auc = roc_auc_score(
        y_test,
        y_prob
    )

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "F1 Score": f1,
        "ROC-AUC": roc_auc
    })


# ==================================================
# RESULTS TABLE
# ==================================================

results_df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("📊 FINAL MODEL PERFORMANCE")
print("=" * 70)

print(
    results_df.to_string(
        index=False,
        formatters={
            "Accuracy": "{:.4f}".format,
            "F1 Score": "{:.4f}".format,
            "ROC-AUC": "{:.4f}".format
        }
    )
)

print("=" * 70)

```

```python
import shap
import matplotlib.pyplot as plt
from pathlib import Path

# ==================================================
# SHAP EXPLAINABILITY
# ==================================================

print("Creating SHAP explanation...")

# Use XGBoost model
explainer = shap.TreeExplainer(xgb_model)

# Use a small sample for faster SHAP calculation
X_shap = X_test_scaled[:100]

# Calculate SHAP values
shap_values = explainer.shap_values(X_shap)

print("✅ SHAP values calculated")
print("SHAP shape:", shap_values.shape)


# ==================================================
# FEATURE NAMES
# ==================================================

feature_names = X.columns.tolist()


# ==================================================
# SCREENSHOTS FOLDER
# ==================================================

SCREENSHOT_DIR = BASE_DIR / "screenshots"

SCREENSHOT_DIR.mkdir(
    exist_ok=True
)


# ==================================================
# SHAP SUMMARY PLOT
# ==================================================

plt.figure()

shap.summary_plot(
    shap_values,
    X_shap,
    feature_names=feature_names,
    show=False
)

plt.tight_layout()

summary_path = (
    SCREENSHOT_DIR
    / "shap_summary.png"
)

plt.savefig(
    summary_path,
    dpi=200,
    bbox_inches="tight"
)

plt.close()

print(
    "✅ SHAP summary saved:",
    summary_path
)


# ==================================================
# SHAP BAR IMPORTANCE
# ==================================================

plt.figure()

shap.summary_plot(
    shap_values,
    X_shap,
    feature_names=feature_names,
    plot_type="bar",
    show=False
)

plt.tight_layout()

bar_path = (
    SCREENSHOT_DIR
    / "shap_feature_importance.png"
)

plt.savefig(
    bar_path,
    dpi=200,
    bbox_inches="tight"
)

plt.close()

print(
    "✅ Feature importance saved:",
    bar_path
)
```
