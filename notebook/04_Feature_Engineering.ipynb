```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/processed/loan_cleaned.csv")

print("Shape:", df.shape)
```

```python
print(df["default"].value_counts())
```

```python
print(df["default"].value_counts(normalize=True) * 100)
```

Create credit-risk features

Feature 1 — Loan to Income

```python
df["loan_to_income"] = (
    df["loan_amnt"] / (df["annual_inc"] + 1)
)
```

Feature 2 — Installment to Income

```python
df["installment_to_income"] = (
    df["installment"] / (df["annual_inc"] + 1)
)
```

Feature 3 — Open Account Ratio

```python
df["open_to_total_accounts"] = (
    df["open_acc"] / (df["total_acc"] + 1)
)
```

Feature 4 — Credit Issue Count

```python
df["credit_issue_count"] = (
    df["delinq_2yrs"] + df["pub_rec"]
)
```

```python
df["revol_bal_to_income"] = (
    df["revol_bal"] / (df["annual_inc"] + 1)
)
```

```python
new_features = [
    "loan_to_income",
    "installment_to_income",
    "open_to_total_accounts",
    "credit_issue_count",
    "revol_bal_to_income"
]

df[new_features].describe().T
```
