```python
# Columns available at the time of loan application/origination
candidate_features = [
    "loan_amnt",
    "term",
    "int_rate",
    "installment",
    "grade",
    "sub_grade",
    "emp_length",
    "home_ownership",
    "annual_inc",
    "verification_status",
    "purpose",
    "addr_state",
    "dti",
    "delinq_2yrs",
    "inq_last_6mths",
    "open_acc",
    "pub_rec",
    "revol_bal",
    "revol_util",
    "total_acc",
    "fico_range_low",
    "fico_range_high",
    "application_type"
]

# Check which of these columns actually exist
available_features = [
    col for col in candidate_features
    if col in df.columns
]

missing_features = [
    col for col in candidate_features
    if col not in df.columns
]

print("Available features:")
print(available_features)

print("\nNot available in dataset:")
print(missing_features)
```

```python
selected_features = [
    "loan_amnt",
    "term",
    "int_rate",
    "installment",
    "grade",
    "sub_grade",
    "emp_length",
    "home_ownership",
    "annual_inc",
    "verification_status",
    "purpose",
    "addr_state",
    "dti",
    "delinq_2yrs",
    "inq_last_6mths",
    "open_acc",
    "pub_rec",
    "revol_bal",
    "revol_util",
    "total_acc",
    "application_type"
]

df_model = df[selected_features + ["loan_status"]].copy()

print("Shape:", df_model.shape)
df_model.head()
```

filter target statuses

```python
target_statuses = [
    "Fully Paid",
    "Charged Off",
    "Default"
]

df_model = df_model[
    df_model["loan_status"].isin(target_statuses)
].copy()

print("Shape after filtering:", df_model.shape)

print("\nLoan Status:")
print(df_model["loan_status"].value_counts())
```

create Target variable 

```python
df_model["default"] = (
    df_model["loan_status"]
    .isin(["Charged Off", "Default"])
    .astype(int)
)
```

## Data Understanding Summary

- Original dataset: 2,260,668 rows and 145 columns.
- Selected modeling dataset contains 23 columns.
- After filtering to resolved loan outcomes, 1,309,638 records remain.
- Target variable: `default`.
- `Fully Paid` is mapped to 0 (No Default).
- `Charged Off` and `Default` are mapped to 1 (Default).
- Default class represents approximately 20.07% of the dataset.
- Four selected features contain missing values.
- Categorical features include loan term, grade, sub-grade, employment length, home ownership, verification status, purpose, state, and application type.
